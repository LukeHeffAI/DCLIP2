```python
def generate_latex_table(json_data):
    # Datasets and model sizes based on LaTeX table
    desc_gen_method = 'gpt4o'
    datasets = ['imagenet', 'cub', 'pets', 'dtd', 'food101', 'places365', 'eurosat']
    # datasets = ['imagenet', 'places365' ]
    model_sizes = ['ViT-B/32', 'ViT-B/16', 'ViT-L/14']
    methods = ['clip', 'e-clip', 'd-clip', 'waffleclip', 'waffleclip+concepts', 'defntaxs']

    # Initialize a structure to store the column values
    column_values = {dataset: {model_size: [] for model_size in model_sizes} for dataset in datasets}

    # Gather values for each column
    for method in methods:
        for dataset in datasets:
            for model_size in model_sizes:
                try:
                    if method in json_data[desc_gen_method][model_size]:
                        result = json_data[desc_gen_method][model_size][method][dataset][f"{method.capitalize()} Accuracy: "]
                    else:
                        result = "-"
                except KeyError:
                    result = "-"
                if isinstance(result, float):
                    column_values[dataset][model_size].append(result)
                else:
                    column_values[dataset][model_size].append(None)

    # Table header formatting
    print("\\begin{table}[ht]")
    print("    \\caption{Comparison of zero-shot visual classification performance across different image classification benchmarks using multiple CLIP backbones.}")
    print("    \\label{tab:clip_comparison}")
    print("    \\begin{center}")
    print("    \\resizebox{\\textwidth}{!}{")
    print("    \\begin{tabular}{lccc|ccc|ccc|ccc|ccc|ccc|ccc|ccc|ccc}")

    print("        \\hline")
    print("        \\multirow{2}{*}{\\textbf{Method}} & \\multicolumn{3}{c}{\\textbf{ImageNet}} & \\multicolumn{3}{c}{\\textbf{CUB}} & \\multicolumn{3}{c}{\\textbf{Oxford Pets}} & \\multicolumn{3}{c}{\\textbf{DTD}} & \\multicolumn{3}{c}{\\textbf{Food101}} & \\multicolumn{3}{c}{\\textbf{Places365}} & \\multicolumn{3}{c}{\\textbf{EuroSAT}} \\\\")
    print("        \\cmidrule(lr){2-4} \\cmidrule(lr){5-7} \\cmidrule(lr){8-10} \\cmidrule(lr){11-13} \\cmidrule(lr){14-16} \\cmidrule(lr){17-19} \\cmidrule(lr){20-22}")
    print("        & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 \\\\")
    print("        \\hline")

    # Iterate over each method and print results in the LaTeX format
    for method in methods:
        row = f"        {method.capitalize()}"  # Start each row with the method name
        for dataset in datasets:
            for model_size in model_sizes:
                # Extract the accuracy for each combination
                try:
                    if method in json_data[desc_gen_method][model_size]:
                        result = json_data[desc_gen_method][model_size][method][dataset][f"{method.capitalize()} Accuracy: "]
                    else:
                        result = "-"
                except KeyError:
                    result = "-"

                if isinstance(result, float):
                    # Determine if the value is the max or second max in the column
                    col_values = column_values[dataset][model_size]
                    max_val = max(v for v in col_values if v is not None)
                    second_max_val = sorted([v for v in col_values if v is not None], reverse=True)[1] if len(col_values) > 1 else None

                    if result == max_val:
                        result_str = f"\\textbf{{{result:.2f}}}"
                    elif result == second_max_val:
                        result_str = f"\\underline{{{result:.2f}}}"
                    else:
                        result_str = f"{result:.2f}"
                else:
                    result_str = result
                row += f" & {result_str}"
        row += " \\\\"
        print(row.replace("Clip", "CLIP").replace("clip","CLIP").replace("Defntaxs", "DefNTaxS"))
    
    # Close the LaTeX table
    print("        \\hline")
    print("    \\end{tabular}")
    print("    }")
    print("    \\end{center}")
    print("\\end{table}")



generate_latex_table(paper_results_data)
```

```python
import json

filename = 'results/all_backbone_method_dataset_experiment_results.json'

with open(filename) as f:
    paper_results_data = json.load(f)

def generate_stats_latex_table(json_data):
    """Generate LaTeX table with mean±SE, per-cell Δ vs DefNTaxS (coloured),
       Win-Count and Near-Win-Count columns, plus swap-difference summary rows."""
    desc_gen_method = 'gpt-3'
    datasets       = ['imagenet','cub','pets','dtd','food101','places365','eurosat']
    model_sizes    = ['ViT-B/32','ViT-B/16','ViT-L/14']
    methods        = ['defntaxs','waffletaxs','taxclip','waffletaxs_swap','taxclip_swap']

    #-------------------------------------------------------------------
    # 1) Gather all results into a data structure
    #-------------------------------------------------------------------
    # results_dict[method][dataset][model_size] = float or None
    results_dict = {
        method: {
            dataset: {model_size: None for model_size in model_sizes}
            for dataset in datasets
        }
        for method in methods
    }

    # For each (method, dataset, model_size), attempt to load from JSON
    for method in methods:
        for dataset in datasets:
            for model_size in model_sizes:
                try:
                    # Check if the method (e.g. 'clip') is in the JSON for this backbone
                    if method in json_data[desc_gen_method][model_size]:
                        # Example: "clip" -> "Clip Accuracy: "
                        accuracy_key = f"{method.capitalize()} Accuracy: "
                        result = json_data[desc_gen_method][model_size][method][dataset][accuracy_key]
                    else:
                        result = None
                except (KeyError, TypeError):
                    result = None

                if isinstance(result, float):
                    results_dict[method][dataset][model_size] = result
                else:
                    results_dict[method][dataset][model_size] = None

    #-------------------------------------------------------------------
    # 2) Compute column-wise info for max/second-max highlighting
    #    (used for bold and underline in the main rows)
    #-------------------------------------------------------------------
    column_values = {
        ds: {ms: [] for ms in model_sizes} for ds in datasets
    }

    for ds in datasets:
        for ms in model_sizes:
            for m in methods:
                val = results_dict[m][ds][ms]
                if isinstance(val, float):
                    column_values[ds][ms].append(val)

    def find_max_and_second_max(values_list):
        valid_vals = sorted(values_list, reverse=True)
        if not valid_vals:
            return (None, None)
        if len(valid_vals) == 1:
            return (valid_vals[0], None)
        return (valid_vals[0], valid_vals[1])

    #-------------------------------------------------------------------
    # 3) Print LaTeX table header with the "Mean" column
    #-------------------------------------------------------------------
    print("\\begin{table}[ht]")
    print("    \\caption{Comparison of zero-shot visual classification performance across different benchmarks using multiple CLIP backbones.}")
    print("    \\label{tab:clip_comparison}")
    print("    \\begin{center}")
    print("    \\resizebox{\\textwidth}{!}{")

    # We add one extra column for “Mean”
    print("    \\begin{tabular}{l" + "ccc" * len(datasets) + "|c}")
    print("        \\hline")

    # First row of header
    print("        \\multirow{2}{*}{\\textbf{Method}}", end="")
    for d in datasets:
        print(f" & \\multicolumn{{3}}{{c}}{{\\textbf{{{d.capitalize()}}}}}", end="")
    print(" & \\multirow{2}{*}{\\textbf{Mean}} \\\\")

    # Middle rule lines for each dataset's 3 columns
    col_start = 2
    for _ in range(len(datasets)):
        col_end = col_start + 2
        print(f"        \\cmidrule(lr){{{col_start}-{col_end}}}", end="")
        col_start += 3
    print("")

    # Second row of header
    print("        ", end="")
    for _ in datasets:
        print(" & B/32 & B/16 & L/14", end="")
    print(" &  -- \\\\")
    print("        \\hline")

    #-------------------------------------------------------------------
    # 4) Print the main rows
    #-------------------------------------------------------------------
    for method in methods:
        row_vals = []
        row_str = f"        {method.capitalize()}"

        for ds in datasets:
            for ms in model_sizes:
                val = results_dict[method][ds][ms]
                if isinstance(val, float):
                    # Determine max/2nd max for this column
                    max_val, second_val = find_max_and_second_max(column_values[ds][ms])
                    if val == max_val:
                        val_str = f"\\textbf{{{val:.2f}}}"
                    elif val == second_val:
                        val_str = f"\\underline{{{val:.2f}}}"
                    else:
                        val_str = f"{val:.2f}"
                    row_vals.append(val)
                else:
                    val_str = "-"
                row_str += f" & {val_str}"

        # Compute mean for this method across all valid entries
        if row_vals:
            mean_val = sum(row_vals) / len(row_vals)
            row_str += f" & {mean_val:.2f}"
        else:
            row_str += " & -"

        # Do final text replacements for nicer method names
        row_str = (
            row_str
            .replace("Clip", "CLIP")  # "clip" -> "CLIP"
            .replace("D-clip", "D-CLIP")
            .replace("Defntaxs", "DefNTaxS")
        )
        print(row_str + " \\\\")

    #-------------------------------------------------------------------
    # 5) Difference rows
    #-------------------------------------------------------------------
    def format_diff(d):
        if d >= 0:
            return f"\\textcolor{{green}}{{+{d:.2f}}}"
        else:
            return f"\\textcolor{{red}}{{{d:.2f}}}"

    def print_difference_row(label, minuend_key, subtrahend_key):
        row_vals = []
        row_str = f"        {label}"
        for ds in datasets:
            for ms in model_sizes:
                val_min = results_dict[minuend_key][ds][ms]
                val_sub = results_dict[subtrahend_key][ds][ms]
                if isinstance(val_min, float) and isinstance(val_sub, float):
                    diff = val_min - val_sub
                    row_vals.append(diff)
                    row_str += f" & {format_diff(diff)}"
                else:
                    row_str += " & -"

        if row_vals:
            mean_diff = sum(row_vals) / len(row_vals)
            row_str += f" & {format_diff(mean_diff)}"
        else:
            row_str += " & -"

        print(row_str + " \\\\")

    # Differences: DefNTaxS - CLIP, DefNTaxS - D-CLIP
    # (Adjust if you only have 'clip' / 'd-clip' in the JSON)
    print_difference_row("DefNTaxS - CLIP", "defntaxs", "clip")
    print_difference_row("DefNTaxS - D-CLIP", "defntaxs", "d-clip")

    #-------------------------------------------------------------------
    # 6) Close out the table
    #-------------------------------------------------------------------
    print("        \\hline")
    print("    \\end{tabular}")
    print("    }")
    print("    \\end{center}")
    print("\\end{table}")

generate_latex_table(paper_results_data)
```

```python
import json
import numpy as np

# Load the multiple runs results data
filename = 'results/multiple_runs_experiment_results.json'

try:
    with open(filename) as f:
        multiple_runs_data = json.load(f)
    print(f"Successfully loaded data from {filename}\n\n")
except FileNotFoundError:
    multiple_runs_data = {}
    print(f"File {filename} not found. Using empty dictionary.")

def generate_stats_latex_table(json_data):
    desc_gen_method = 'gpt-3'
    datasets       = ['imagenet','cub','pets','dtd','food101','places365','eurosat']
    model_sizes    = ['ViT-B/32','ViT-B/16','ViT-L/14']
    methods        = ['defntaxs','waffletaxs','taxclip','waffletaxs_swap','taxclip_swap']

    # 1) initialize container
    stats_values = {
        d: {sz: {m:{'mean':None,'std':None} for m in methods}
            for sz in model_sizes}
        for d in datasets
    }

    # 2) fill in mean±std
    for method in methods:
        # pick the JSON key for this method
        for d in datasets:
            for sz in model_sizes:
                try:
                    runs = json_data[desc_gen_method][sz][method][d]
                except KeyError:
                    continue

                # determine which Top-1 Accuracy field to pull
                if method in ('clip','e-clip','d-clip'):
                    key = 'CLIP Top-1 Accuracy'
                else:
                    key = f"{method.capitalize()} Top-1 Accuracy"

                # collect all runs that actually have that key
                vals = [run[key] for run in runs if key in run]
                if not vals:
                    continue

                stats_values[d][sz][method]['mean'] = np.mean(vals)
                stats_values[d][sz][method]['std']  = np.std(vals)

    # --- 3) compute wins & near‑wins per cell ---
    wins     = {m:0 for m in methods}
    nearwins = {m:0 for m in methods}
    for d in datasets:
        for sz in model_sizes:
            cell = stats_values[d][sz]
            valid = {m:cell[m]['mean'] for m in methods if cell[m]['mean'] is not None}
            if not valid: continue
            max_mean = max(valid.values())
            winners = [m for m,v in valid.items() if abs(v-max_mean)<1e-8]
            for w in winners:
                wins[w] += 1
            win_se = max(cell[w]['std'] for w in winners)
            for m,v in valid.items():
                if v >= max_mean - win_se:
                    nearwins[m] += 1

    # --- 4) LaTeX header ---
    col_spec = 'l' + '|ccc'*len(datasets) + '|cc'
    print(r"\begin{table*}[ht]")
    print("\t" + r"\caption{Comparison of DefNTaxS and two random character/word-augmented approaches, WaffleTaxS and TaxCLIP, for zero-shot visual classification performance across 5 iterations, displaying the mean and standard error. An augmentation of this approach swaps the order of the placement of the descriptors and subcategories in the text prompt, to account for the effect of reducing contextual attention throughout the CLIP context window, referred to as WaffleTaxS$_{swap}$ and TaxCLIP$_{swap}$, respectively.}")
    print("\t" + r"\label{tab:stats_comparison1+swaps}")
    print("\t" + r"\centering")
    print("\t" + r"\resizebox{\textwidth}{!}{")
    print("\t" + f"\\begin{{tabular}}{{{col_spec}}}")
    print("\t\t" + r"\toprule")
    hdr1 = "\t\t" + r"\multirow{2}{*}{\textbf{Method}} "
    for d in datasets:
        hdr1 += f"& \\multicolumn{{3}}{{c}}{{\\textbf{{{d.capitalize()}}}}} "
    hdr1 += "& \\textbf{Wins} & \\textbf{Near-Wins} \\"
    print(hdr1)
    # cmidrule
    cuts = []
    start = 2
    for _ in datasets:
        cuts.append(f"{start}-{start+2}")
        start += 3
    cm = "\t\t" + "\\cmidrule(lr){" + "} \\cmidrule(lr){".join(cuts) + "}"
    print(cm)
    sizes = "& " + " & ".join(["B/32 & B/16 & L/14"]*len(datasets)) + " & -- & -- \\\\"
    print("\t\t" + sizes)
    print("\t\t" + r"\midrule")

    # --- 5) print rows ---
    for method in methods:
        nice = method.replace("defntaxs","DefNTaxS") \
                      .replace("waffletaxs","WaffleTaxS") \
                      .replace("taxclip","TaxCLIP") \
                      .replace("_swap", r"$_\mathit{swap}$")
        row = f"{nice}"
        for d in datasets:
            for sz in model_sizes:
                s = stats_values[d][sz][method]
                if s['mean'] is None:
                    row += " & -"
                else:
                    row += f" & {s['mean']:.2f} $\\pm$ {s['std']:.2f}"
        row += f" & {wins[method]} & {nearwins[method]} \\\\"
        print("\t\t" + row)

        if method != 'defntaxs':
            delta = "$\\Delta$ vs~DefNTaxS"
            for d in datasets:
                for sz in model_sizes:
                    base = stats_values[d][sz]['defntaxs']['mean']
                    cur  = stats_values[d][sz][method]['mean']
                    if base is None or cur is None:
                        delta += " & -"
                    else:
                        diff = cur-base
                        colour = "blue" if diff>=0 else "red"
                        delta += f" & \\textcolor{{{colour}}}{{{diff:+.2f}}}"
            delta += " & -- & -- \\\\"
            print("\t\t" + delta)

    # --- 6) swap diff summaries ---
    print("\t\t" + r"\midrule")
    for orig, swap, lbl in [
        ('waffletaxs','waffletaxs_swap', r'WaffleTaxS$_\mathit{swap}$ vs WaffleTaxS'),
        ('taxclip',  'taxclip_swap',   r'TaxCLIP$_\mathit{swap}$ vs TaxCLIP')
    ]:
        sd = f"$\\Delta$ {lbl}"
        for d in datasets:
            for sz in model_sizes:
                o = stats_values[d][sz][orig]['mean']
                s = stats_values[d][sz][swap]['mean']
                if o is None or s is None:
                    sd += " & -"
                else:
                    diff = s-o
                    col  = "blue" if diff>=0 else "red"
                    sd += f" & \\textcolor{{{col}}}{{{diff:+.2f}}}"
        sd += " & -- & -- \\\\"
        print("\t\t" + sd)

    print("\t\t" + r"\bottomrule")
    print("\t" + r"\end{tabular}")
    print("\t" + r"}")
    print(r"\end{table*}")

# --- call the function to print LaTeX ---
generate_stats_latex_table(multiple_runs_data)
```
