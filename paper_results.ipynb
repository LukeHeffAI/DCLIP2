```python
import json

filename = 'results/NEW_all_backbone_method_dataset_experiment_results.json'

with open(filename) as f:
    paper_results_data = json.load(f)

def generate_latex_table(json_data):
    # << 1) Fix the descriptor key to match JSON >>
    desc_gen_method = 'gpt-3'  # Matches your JSON: {"gpt-3": {...}}

    # List the datasets and model sizes you want in the table
    datasets = ['imagenet', 'cub', 'pets', 'dtd', 'food101', 'places365', 'eurosat']
    model_sizes = ['ViT-B/32', 'ViT-B/16', 'ViT-L/14']

    # Methods you want to compare; keep in mind only "clip" and "d-clip"
    # actually appear in your snippet. Others (defntaxs, etc.) may remain "-"
    methods = ['clip', 'e-clip', 'd-clip', 'waffleclip', 'waffleclip+concepts', 'defntaxs']

    #-------------------------------------------------------------------
    # 1) Gather all results into a data structure
    #-------------------------------------------------------------------
    # results_dict[method][dataset][model_size] = float or None
    results_dict = {
        method: {
            dataset: {model_size: None for model_size in model_sizes}
            for dataset in datasets
        }
        for method in methods
    }

    # For each (method, dataset, model_size), attempt to load from JSON
    for method in methods:
        for dataset in datasets:
            for model_size in model_sizes:
                try:
                    # Check if the method (e.g. 'clip') is in the JSON for this backbone
                    if method in json_data[desc_gen_method][model_size]:
                        # Example: "clip" -> "Clip Accuracy: "
                        accuracy_key = f"{method.capitalize()} Accuracy: "
                        result = json_data[desc_gen_method][model_size][method][dataset][accuracy_key]
                    else:
                        result = None
                except (KeyError, TypeError):
                    result = None

                if isinstance(result, float):
                    results_dict[method][dataset][model_size] = result
                else:
                    results_dict[method][dataset][model_size] = None

    #-------------------------------------------------------------------
    # 2) Compute column-wise info for max/second-max highlighting
    #    (used for bold and underline in the main rows)
    #-------------------------------------------------------------------
    column_values = {
        ds: {ms: [] for ms in model_sizes} for ds in datasets
    }

    for ds in datasets:
        for ms in model_sizes:
            for m in methods:
                val = results_dict[m][ds][ms]
                if isinstance(val, float):
                    column_values[ds][ms].append(val)

    def find_max_and_second_max(values_list):
        valid_vals = sorted(values_list, reverse=True)
        if not valid_vals:
            return (None, None)
        if len(valid_vals) == 1:
            return (valid_vals[0], None)
        return (valid_vals[0], valid_vals[1])

    #-------------------------------------------------------------------
    # 3) Print LaTeX table header with the "Mean" column
    #-------------------------------------------------------------------
    print("\\begin{table}[ht]")
    print("    \\caption{Comparison of zero-shot visual classification performance across different benchmarks using multiple CLIP backbones.}")
    print("    \\label{tab:clip_comparison}")
    print("    \\begin{center}")
    print("    \\resizebox{\\textwidth}{!}{")

    # We add one extra column for “Mean”
    print("    \\begin{tabular}{l" + "ccc" * len(datasets) + "|c}")
    print("        \\hline")

    # First row of header
    print("        \\multirow{2}{*}{\\textbf{Method}}", end="")
    for d in datasets:
        print(f" & \\multicolumn{{3}}{{c}}{{\\textbf{{{d.capitalize()}}}}}", end="")
    print(" & \\multirow{2}{*}{\\textbf{Mean}} \\\\")

    # Middle rule lines for each dataset's 3 columns
    col_start = 2
    for _ in range(len(datasets)):
        col_end = col_start + 2
        print(f"        \\cmidrule(lr){{{col_start}-{col_end}}}", end="")
        col_start += 3
    print("")

    # Second row of header
    print("        ", end="")
    for _ in datasets:
        print(" & B/32 & B/16 & L/14", end="")
    print(" &  -- \\\\")
    print("        \\hline")

    #-------------------------------------------------------------------
    # 4) Print the main rows
    #-------------------------------------------------------------------
    for method in methods:
        row_vals = []
        row_str = f"        {method.capitalize()}"

        for ds in datasets:
            for ms in model_sizes:
                val = results_dict[method][ds][ms]
                if isinstance(val, float):
                    # Determine max/2nd max for this column
                    max_val, second_val = find_max_and_second_max(column_values[ds][ms])
                    if val == max_val:
                        val_str = f"\\textbf{{{val:.2f}}}"
                    elif val == second_val:
                        val_str = f"\\underline{{{val:.2f}}}"
                    else:
                        val_str = f"{val:.2f}"
                    row_vals.append(val)
                else:
                    val_str = "-"
                row_str += f" & {val_str}"

        # Compute mean for this method across all valid entries
        if row_vals:
            mean_val = sum(row_vals) / len(row_vals)
            row_str += f" & {mean_val:.2f}"
        else:
            row_str += " & -"

        # Do final text replacements for nicer method names
        row_str = (
            row_str
            .replace("Clip", "CLIP")  # "clip" -> "CLIP"
            .replace("D-clip", "D-CLIP")
            .replace("Defntaxs", "DefNTaxS")
        )
        print(row_str + " \\\\")

    #-------------------------------------------------------------------
    # 5) Difference rows
    #-------------------------------------------------------------------
    def format_diff(d):
        if d >= 0:
            return f"\\textcolor{{green}}{{+{d:.2f}}}"
        else:
            return f"\\textcolor{{red}}{{{d:.2f}}}"

    def print_difference_row(label, minuend_key, subtrahend_key):
        row_vals = []
        row_str = f"        {label}"
        for ds in datasets:
            for ms in model_sizes:
                val_min = results_dict[minuend_key][ds][ms]
                val_sub = results_dict[subtrahend_key][ds][ms]
                if isinstance(val_min, float) and isinstance(val_sub, float):
                    diff = val_min - val_sub
                    row_vals.append(diff)
                    row_str += f" & {format_diff(diff)}"
                else:
                    row_str += " & -"

        if row_vals:
            mean_diff = sum(row_vals) / len(row_vals)
            row_str += f" & {format_diff(mean_diff)}"
        else:
            row_str += " & -"

        print(row_str + " \\\\")

    # Differences: DefNTaxS - CLIP, DefNTaxS - D-CLIP
    # (Adjust if you only have 'clip' / 'd-clip' in the JSON)
    print_difference_row("DefNTaxS - CLIP", "defntaxs", "clip")
    print_difference_row("DefNTaxS - D-CLIP", "defntaxs", "d-clip")

    #-------------------------------------------------------------------
    # 6) Close out the table
    #-------------------------------------------------------------------
    print("        \\hline")
    print("    \\end{tabular}")
    print("    }")
    print("    \\end{center}")
    print("\\end{table}")

generate_latex_table(paper_results_data)
```

```python
import json

filename = 'results/NEW_all_backbone_method_dataset_experiment_results_merge.json'

with open(filename) as f:
    paper_results_data = json.load(f)

def generate_latex_table(json_data):
    # << 1) Fix the descriptor key to match JSON >>
    desc_gen_method = 'gpt-3'  # Matches your JSON: {"gpt-3": {...}}

    # List the datasets and model sizes you want in the table
    datasets = ['imagenet', 'cub', 'pets', 'dtd', 'food101', 'places365', 'eurosat']
    model_sizes = ['ViT-B/32']

    # Methods you want to compare; keep in mind only "clip" and "d-clip"
    # actually appear in your snippet. Others (defntaxs, etc.) may remain "-"
    # methods = ['clip', 'e-clip', 'd-clip', 'waffleclip', 'waffleclip+concepts', 'defntaxs']
    methods = ['clip', 'e-clip', 'd-clip', 'waffleclip', 'waffleclip+concepts', 'cupl', 'cgptp', 'chils_no_rw', 'defntaxs']

    #-------------------------------------------------------------------
    # 1) Gather all results into a data structure
    #-------------------------------------------------------------------
    # results_dict[method][dataset][model_size] = float or None
    results_dict = {
        method: {
            dataset: {model_size: None for model_size in model_sizes}
            for dataset in datasets
        }
        for method in methods
    }

    # For each (method, dataset, model_size), attempt to load from JSON
    for method in methods:
        for dataset in datasets:
            for model_size in model_sizes:
                try:
                    # Check if the method (e.g. 'clip') is in the JSON for this backbone
                    if method in json_data[desc_gen_method][model_size]:
                        # Example: "clip" -> "Clip Accuracy: "
                        accuracy_key = f"{method.capitalize()} Accuracy: "
                        result = json_data[desc_gen_method][model_size][method][dataset][accuracy_key]
                    else:
                        result = None
                except (KeyError, TypeError):
                    result = None

                if isinstance(result, float):
                    results_dict[method][dataset][model_size] = result
                else:
                    results_dict[method][dataset][model_size] = None

    #-------------------------------------------------------------------
    # 2) Compute column-wise info for max/second-max highlighting
    #    (used for bold and underline in the main rows)
    #-------------------------------------------------------------------
    column_values = {
        ds: {ms: [] for ms in model_sizes} for ds in datasets
    }
    for ds in datasets:
        for ms in model_sizes:
            for m in methods:
                val = results_dict[m][ds][ms]
                if isinstance(val, float):
                    column_values[ds][ms].append(val)

    def find_max_and_second_max(values_list):
        valid_vals = sorted(values_list, reverse=True)
        if not valid_vals:
            return (None, None)
        if len(valid_vals) == 1:
            return (valid_vals[0], None)
        return (valid_vals[0], valid_vals[1])

    #-------------------------------------------------------------------
    # 3) Print LaTeX table header with the separate Mean columns for each backbone
    #-------------------------------------------------------------------
    print(r"\begin{table*}[ht]")
    print("    \\caption{Comparison of zero-shot visual classification performance across different benchmarks using multiple CLIP backbones.}")
    print(r"    \label{tab:label} % give label")
    print("    \\centering")
    print("    \\resizebox{\\textwidth}{!}{")
    # Now the tabular has 1 (Method) + len(datasets)*3 (each dataset has 3 columns) + 3 (one per backbone mean)
    cols = "".join("c" * len(model_sizes))
    print("    \\begin{tabular}{|l|" + f"{cols}|" * len(datasets) + f"{cols}|" + "}")
    print("        \\hline")
    
    # First header row: print dataset names and a multicolumn for the mean values
    print("        \\textbf{Method}", end="")
    for d in datasets:
        d = ( 
            d.replace("imagenet", "ImageNet")
            .replace("dtd", "DTD")
            .replace("cub", "CUB")
            .replace("euroSAT", "EuroSAT"))
        

        print(f" & \\multicolumn{{{len(model_sizes)}}}{{c}}{{\\textbf{{{d.capitalize()}}}}}", end="")
    print(f" & \\multicolumn{{{len(model_sizes)}}}{{c}}{{\\textbf{{Mean}}}} \\\\")

    # Draw cmidrules for each dataset column group
    col_start = 2
    for _ in range(len(datasets)):
        # if len(model_sizes) == 1:
        #     col_end = ""
        # else:
        col_end = str(col_start + len(model_sizes) - 1)
        print(f"        \\cline(lr){{{col_start}-{col_end}}}", end="")
        col_start += len(model_sizes)
    # Add cmidrule for the mean columns
    mean_start = 1 + len(datasets) * 3 + 1  # equals len(datasets)*3 + 2
    mean_end = mean_start + 3 - 1
    print(f"        \\cline(lr){{{mean_start}-{mean_end}}}")
    
    # Second header row: print the small column headings for each dataset and the backbone names for mean columns
    table_backbones = " & ".join(model_sizes)
    print("", end="")
    for _ in datasets:
        print(f"        & {table_backbones}\n", end="")
    print(f"        & {table_backbones} \\\\")
    print("        \\hline")
    
    #-------------------------------------------------------------------
    # 4) Print the main rows (one per method) with means computed per backbone.
    #-------------------------------------------------------------------
    for method in methods:
        row_str = f"        {method.capitalize()}"
        # Initialize a dictionary to store valid values for the mean per backbone
        mean_per_backbone = {ms: [] for ms in model_sizes}

        for ds in datasets:
            for ms in model_sizes:
                val = results_dict[method][ds][ms]
                if isinstance(val, float):
                    # Determine max/2nd max for this column
                    max_val, second_val = find_max_and_second_max(column_values[ds][ms])
                    if val == max_val:
                        val_str = f"\\textbf{{{val:.2f}}}"
                    elif val == second_val:
                        val_str = f"\\underline{{{val:.2f}}}"
                    else:
                        val_str = f"{val:.2f}"
                    mean_per_backbone[ms].append(val)
                else:
                    val_str = "-"
                row_str += f" & {val_str}"

        # For each backbone, compute the mean across datasets
        for ms in model_sizes:
            vals = mean_per_backbone[ms]
            if vals:
                mean_val = sum(vals) / len(vals)
                row_str += f" & {mean_val:.2f}"
            else:
                row_str += " & -"

        # Do final text replacements for nicer method names
        row_str = (
            row_str
            .replace("Clip", "CLIP")
            .replace("E-clip", "E-CLIP")
            .replace("D-clip", "D-CLIP")
            .replace("Waffleclip", "WaffleCLIP")
            .replace("Waffleclip+concepts", "WaffleCLIP+Concepts")
            .replace("Chils", "CHiLS")
            .replace("_superclass", " Superclass")
            .replace("_no_rw", "")
            .replace("Cupl", "CUPL")
            .replace("Cgptp", "CGPT-Powered")
            .replace("Defntaxs", "DefNTaxS")
        )
        print(row_str + " \\\\")
    
    #-------------------------------------------------------------------
    # 5) Difference rows (computing mean difference per backbone)
    #-------------------------------------------------------------------
    # def format_diff(d):
    #     if d >= 0:
    #         return f"\\textcolor{{green}}{{+{d:.2f}}}"
    #     else:
    #         return f"\\textcolor{{red}}{{{d:.2f}}}"

    def format_diff(d):
        if d >= 0:
            return f"{{{d:.2f}}}"
        else:
            return f"{{{d:.2f}}}"

    def print_difference_row(label, minuend_key, subtrahend_key):
        row_str = f"        {label}"
        # Initialize dictionary to store differences for each backbone
        mean_diff_backbone = {ms: [] for ms in model_sizes}
        for ds in datasets:
            for ms in model_sizes:
                val_min = results_dict[minuend_key][ds][ms]
                val_sub = results_dict[subtrahend_key][ds][ms]
                if isinstance(val_min, float) and isinstance(val_sub, float):
                    diff = val_min - val_sub
                    mean_diff_backbone[ms].append(diff)
                    row_str += f" & {format_diff(diff)}"
                else:
                    row_str += " & -"
        # Compute and append the mean differences per backbone
        for ms in model_sizes:
            diffs = mean_diff_backbone[ms]
            if diffs:
                mean_diff = sum(diffs) / len(diffs)
                row_str += f" & {format_diff(mean_diff)}"
            else:
                row_str += " & -"
        print(row_str + " \\\\")
    
    # Differences: DefNTaxS - CLIP, DefNTaxS - D-CLIP
    # (Adjust if you only have 'clip' / 'd-clip' in the JSON)
    print_difference_row("DefNTaxS - CLIP", "defntaxs", "clip")
    print_difference_row("DefNTaxS - D-CLIP", "defntaxs", "d-clip")
    
    #-------------------------------------------------------------------
    # 6) Close out the table
    #-------------------------------------------------------------------
    print("        \\hline")
    print("    \\end{tabular}")
    print("    }")
    print(r"\end{table*}")

generate_latex_table(paper_results_data)
```

```python
import os
import json

def load_json_from(filename):
    with open(filename) as f:
        return json.load(f)
    
def save_json_to(filename, data):
    with open(filename, 'w') as f:
        json.dump(data, f, indent=4)

def latex_to_list(latex_str):
    # remove the leading cell (e.g. \cite{...} or method name) then split on &
    body = latex_str.split(" & ", 1)[1]
    # strip LaTeX commands and braces
    for cmd in (r"\\", "{", "}", "cite", r"\underline", r"\textbf"):
        body = body.replace(cmd, "")
    # split into individual string values and strip whitespace
    parts = [p.strip() for p in body.split("&")]
    # drop any trailing empty cells (e.g. before the final \\)
    return [p for p in parts if p]

def extract_values_from(target_method):
    latex_cgptp = r"\cite{ren2024chatgpt} & \underline{63.32} & \underline{68.03} & 74.55 & \underline{53.72} & \underline{57.81} & 63.89 & 82.90 & 86.16 & 91.77 & \underline{45.81} & \underline{47.18} & \textbf{56.05} & 80.98 & 85.39 & 90.49 & 39.91 & \underline{41.02} & \underline{41.61} & 43.11 & 47.60 & 46.01"
    latex_cupl   = r"CuPL & 62.12 & 66.01 & 73.68 & 52.34 & 56.84 & 63.03 & 81.78 & 84.03 & 84.60 & 40.95 & 42.61 & 43.87 & 79.84 & 83.89 & 88.97 & 38.87 & 39.01 & 38.57 & 41.50 & 38.57 & 48.25"

    if target_method == "cgptp":
        latex_string = latex_cgptp
    elif target_method == "cupl":
        latex_string = latex_cupl
    else:
        raise ValueError(f"Unknown method {target_method!r}")
    return latex_to_list(latex_string)

def process_string_and_values_to(data_dict, model_sizes, datasets, methods):
    """
    Fills in data_dict["gpt-3"][model_size][method][dataset] = {
        "<Method> Accuracy: ": float(...),
        "CLIP Accuracy: ": "clip_baseline"
    }
    """
    # ensure top‐level key
    data_dict.setdefault("gpt-3", {})

    for ms in model_sizes:
        data_dict["gpt-3"].setdefault(ms, {})

    for method in methods:
        values = extract_values_from(method)
        expected = len(model_sizes) * len(datasets)
        if len(values) != expected:
            raise ValueError(
                f"Method {method!r}: expected {expected} values "
                f"(got {len(values)})"
            )

        # fill in model_size × dataset
        for i, ds in enumerate(datasets):
            for j, ms in enumerate(model_sizes):
                idx = i * len(model_sizes) + j
                val = float(values[idx])
                data_dict["gpt-3"][ms].setdefault(method, {})
                data_dict["gpt-3"][ms][method][ds] = {
                    f"{method.capitalize()} Accuracy: ": val,
                    "CLIP Accuracy: ": "clip_baseline"
                }

    return data_dict

def process_json_files(input_file, output_file, model_sizes, datasets, methods):
    # load existing or start fresh
    if os.path.exists(output_file):
        processed = load_json_from(output_file)
        print(f"Loaded existing results from {output_file}")
    else:
        processed = {}

    raw = load_json_from(input_file)
    # merge raw→processed
    merged = process_string_and_values_to(processed, model_sizes, datasets, methods)
    save_json_to(output_file, merged)
    print(f"Wrote merged results to {output_file}")

def main():
    model_sizes = ['ViT-B/32', 'ViT-B/16', 'ViT-L/14']
    datasets = ['imagenet', 'cub', 'pets', 'dtd', 'food101', 'places365', 'eurosat']
    methods  = ['cgptp', 'cupl']

    input_path  = 'results/NEW_all_backbone_method_dataset_experiment_results.json'
    output_path = 'results/NEW_all_backbone_method_dataset_experiment_results_merge.json'

    process_json_files(input_path, output_path, model_sizes, datasets, methods)

if __name__ == "__main__":
    main()
```
