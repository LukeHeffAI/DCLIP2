```python
import json

filename = 'results/all_backbone_method_dataset_experiment_results.json'

with open(filename) as f:
    paper_results_data = json.load(f)

def generate_latex_table(json_data):
    # << 1) Fix the descriptor key to match JSON >>
    desc_gen_method = 'gpt-3'  # Matches your JSON: {"gpt-3": {...}}

    # List the datasets and model sizes you want in the table
    datasets = ['imagenet', 'cub', 'pets', 'dtd', 'food101', 'places365', 'eurosat']
    model_sizes = ['ViT-B/32', 'ViT-B/16', 'ViT-L/14']

    # Methods you want to compare; keep in mind only "clip" and "d-clip"
    # actually appear in your snippet. Others (defntaxs, etc.) may remain "-"
    methods = ['clip', 'e-clip', 'd-clip', 'waffleclip', 'waffleclip+concepts', 'defntaxs']

    #-------------------------------------------------------------------
    # 1) Gather all results into a data structure
    #-------------------------------------------------------------------
    # results_dict[method][dataset][model_size] = float or None
    results_dict = {
        method: {
            dataset: {model_size: None for model_size in model_sizes}
            for dataset in datasets
        }
        for method in methods
    }

    # For each (method, dataset, model_size), attempt to load from JSON
    for method in methods:
        for dataset in datasets:
            for model_size in model_sizes:
                try:
                    # Check if the method (e.g. 'clip') is in the JSON for this backbone
                    if method in json_data[desc_gen_method][model_size]:
                        # Example: "clip" -> "Clip Accuracy: "
                        accuracy_key = f"{method.capitalize()} Accuracy: "
                        result = json_data[desc_gen_method][model_size][method][dataset][accuracy_key]
                    else:
                        result = None
                except (KeyError, TypeError):
                    result = None

                if isinstance(result, float):
                    results_dict[method][dataset][model_size] = result
                else:
                    results_dict[method][dataset][model_size] = None

    #-------------------------------------------------------------------
    # 2) Compute column-wise info for max/second-max highlighting
    #    (used for bold and underline in the main rows)
    #-------------------------------------------------------------------
    column_values = {
        ds: {ms: [] for ms in model_sizes} for ds in datasets
    }

    for ds in datasets:
        for ms in model_sizes:
            for m in methods:
                val = results_dict[m][ds][ms]
                if isinstance(val, float):
                    column_values[ds][ms].append(val)

    def find_max_and_second_max(values_list):
        valid_vals = sorted(values_list, reverse=True)
        if not valid_vals:
            return (None, None)
        if len(valid_vals) == 1:
            return (valid_vals[0], None)
        return (valid_vals[0], valid_vals[1])

    #-------------------------------------------------------------------
    # 3) Print LaTeX table header with the "Mean" column
    #-------------------------------------------------------------------
    print("\\begin{table}[ht]")
    print("    \\caption{Comparison of zero-shot visual classification performance across different benchmarks using multiple CLIP backbones.}")
    print("    \\label{tab:clip_comparison}")
    print("    \\begin{center}")
    print("    \\resizebox{\\textwidth}{!}{")

    # We add one extra column for “Mean”
    print("    \\begin{tabular}{l" + "ccc" * len(datasets) + "|c}")
    print("        \\hline")

    # First row of header
    print("        \\multirow{2}{*}{\\textbf{Method}}", end="")
    for d in datasets:
        print(f" & \\multicolumn{{3}}{{c}}{{\\textbf{{{d.capitalize()}}}}}", end="")
    print(" & \\multirow{2}{*}{\\textbf{Mean}} \\\\")

    # Middle rule lines for each dataset's 3 columns
    col_start = 2
    for _ in range(len(datasets)):
        col_end = col_start + 2
        print(f"        \\cmidrule(lr){{{col_start}-{col_end}}}", end="")
        col_start += 3
    print("")

    # Second row of header
    print("        ", end="")
    for _ in datasets:
        print(" & B/32 & B/16 & L/14", end="")
    print(" &  -- \\\\")
    print("        \\hline")

    #-------------------------------------------------------------------
    # 4) Print the main rows
    #-------------------------------------------------------------------
    for method in methods:
        row_vals = []
        row_str = f"        {method.capitalize()}"

        for ds in datasets:
            for ms in model_sizes:
                val = results_dict[method][ds][ms]
                if isinstance(val, float):
                    # Determine max/2nd max for this column
                    max_val, second_val = find_max_and_second_max(column_values[ds][ms])
                    if val == max_val:
                        val_str = f"\\textbf{{{val:.2f}}}"
                    elif val == second_val:
                        val_str = f"\\underline{{{val:.2f}}}"
                    else:
                        val_str = f"{val:.2f}"
                    row_vals.append(val)
                else:
                    val_str = "-"
                row_str += f" & {val_str}"

        # Compute mean for this method across all valid entries
        if row_vals:
            mean_val = sum(row_vals) / len(row_vals)
            row_str += f" & {mean_val:.2f}"
        else:
            row_str += " & -"

        # Do final text replacements for nicer method names
        row_str = (
            row_str
            .replace("Clip", "CLIP")  # "clip" -> "CLIP"
            .replace("D-clip", "D-CLIP")
            .replace("Defntaxs", "DefNTaxS")
        )
        print(row_str + " \\\\")

    #-------------------------------------------------------------------
    # 5) Difference rows
    #-------------------------------------------------------------------
    def format_diff(d):
        if d >= 0:
            return f"\\textcolor{{green}}{{+{d:.2f}}}"
        else:
            return f"\\textcolor{{red}}{{{d:.2f}}}"

    def print_difference_row(label, minuend_key, subtrahend_key):
        row_vals = []
        row_str = f"        {label}"
        for ds in datasets:
            for ms in model_sizes:
                val_min = results_dict[minuend_key][ds][ms]
                val_sub = results_dict[subtrahend_key][ds][ms]
                if isinstance(val_min, float) and isinstance(val_sub, float):
                    diff = val_min - val_sub
                    row_vals.append(diff)
                    row_str += f" & {format_diff(diff)}"
                else:
                    row_str += " & -"

        if row_vals:
            mean_diff = sum(row_vals) / len(row_vals)
            row_str += f" & {format_diff(mean_diff)}"
        else:
            row_str += " & -"

        print(row_str + " \\\\")

    # Differences: DefNTaxS - CLIP, DefNTaxS - D-CLIP
    # (Adjust if you only have 'clip' / 'd-clip' in the JSON)
    print_difference_row("DefNTaxS - CLIP", "defntaxs", "clip")
    print_difference_row("DefNTaxS - D-CLIP", "defntaxs", "d-clip")

    #-------------------------------------------------------------------
    # 6) Close out the table
    #-------------------------------------------------------------------
    print("        \\hline")
    print("    \\end{tabular}")
    print("    }")
    print("    \\end{center}")
    print("\\end{table}")

generate_latex_table(paper_results_data)
```

```python
import json
import numpy as np

# Load the multiple runs results data
filename = 'results/multiple_runs_experiment_results.json'

try:
    with open(filename) as f:
        multiple_runs_data = json.load(f)
    # print(f"Successfully loaded data from {filename}\n\n")
except FileNotFoundError:
    multiple_runs_data = {}
    print(f"File {filename} not found. Using empty dictionary.")

def generate_stats_latex_table(json_data):
    """Generate LaTeX table with mean and standard deviation values from multiple runs."""
    # Configuration parameters
    desc_gen_method = 'gpt-3'
    datasets = ['imagenet', 'cub', 'pets', 'dtd', 'food101', 'places365', 'eurosat']
    model_sizes = ['ViT-B/32', 'ViT-B/16', 'ViT-L/14']
    # methods = ['clip', 'e-clip', 'd-clip', 'waffleclip', 'waffleclip+concepts', 'defntaxs', 'waffletaxs']
    methods = ['clip', 'defntaxs', 'waffletaxs', 'taxclip']

    # Initialize data structures for statistics
    stats_values = {
        dataset: {
            model_size: {
                method: {'mean': None, 'std': None} for method in methods
            } for model_size in model_sizes
        } for dataset in datasets
    }
    
    # Process the data to compute mean and standard deviation
    for method in methods:
        for dataset in datasets:
            for model_size in model_sizes:
                try:
                    if (desc_gen_method in json_data and 
                        model_size in json_data[desc_gen_method] and 
                        method in json_data[desc_gen_method][model_size] and
                        dataset in json_data[desc_gen_method][model_size][method]):
                        
                        # Get all runs for this configuration
                        runs = json_data[desc_gen_method][model_size][method][dataset]
                        if runs:
                            # Extract accuracy values from all runs
                            method_key = f"{method.capitalize()} Top-1 Accuracy"
                            if method == 'clip' or method == 'e-clip' or method == 'd-clip':
                                method_key = "CLIP Top-1 Accuracy"
                                
                            accuracy_values = [run.get(method_key) for run in runs if method_key in run]
                            
                            if accuracy_values:
                                # Calculate mean and std
                                mean_acc = np.mean(accuracy_values)
                                std_acc = np.std(accuracy_values)
                                
                                # Store the statistics
                                stats_values[dataset][model_size][method]['mean'] = mean_acc
                                stats_values[dataset][model_size][method]['std'] = std_acc
                except Exception as e:
                    print(f"Error processing {method}, {dataset}, {model_size}: {e}")
    
    # Generate the LaTeX table
    print("\\begin{table}[ht]")
    print("    \\caption{Mean and standard error of mean of zero-shot visual classification performance across multiple runs.}")
    print("    \\label{tab:stats_comparison}")
    print("    \\begin{center}")
    print("    \\resizebox{\\textwidth}{!}{")
    print("    \\begin{tabular}{lccc|ccc|ccc|ccc|ccc|ccc|ccc|ccc}")
    
    print("        \\hline")
    print("        \\multirow{2}{*}{\\textbf{Method}} & \\multicolumn{3}{c}{\\textbf{ImageNet}} & \\multicolumn{3}{c}{\\textbf{CUB}} & \\multicolumn{3}{c}{\\textbf{Oxford Pets}} & \\multicolumn{3}{c}{\\textbf{DTD}} & \\multicolumn{3}{c}{\\textbf{Food101}} & \\multicolumn{3}{c}{\\textbf{Places365}} & \\multicolumn{3}{c}{\\textbf{EuroSAT}} \\\\")
    print("        \\cmidrule(lr){2-4} \\cmidrule(lr){5-7} \\cmidrule(lr){8-10} \\cmidrule(lr){11-13} \\cmidrule(lr){14-16} \\cmidrule(lr){17-19} \\cmidrule(lr){20-22}")
    print("        & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 \\\\")
    print("        \\hline")
    
    # Format and print each method row
    for method in methods:
        nice_method_name = (
            method
            .capitalize()
            .replace("clip", "CLIP")
            .replace("Clip", "CLIP")
            .replace("e-", "E-").replace("d-", "D-")
            .replace("waffleclip", "WaffleCLIP")
            .replace("waffleclip+concepts", "WaffleCLIP+Concepts")
            .replace("defntaxs", "DefNTaxS")
            .replace("waffletaxs", "WaffleTaxS")
        )
        row = f"        {nice_method_name}"
        
        for dataset in datasets:
            for model_size in model_sizes:
                stats = stats_values[dataset][model_size][method]
                
                if stats['mean'] is not None and stats['std'] is not None:
                    # Format with mean ± std
                    row += f" & {stats['mean']:.2f} $\\pm$ {stats['std']:.2f}"
                else:
                    row += " & -"
        
        print(row + " \\\\")
    
    # Close the LaTeX table
    print("        \\hline")
    print("    \\end{tabular}")
    print("    }")
    print("    \\end{center}")
    print("\\end{table}")

# Generate the table
generate_stats_latex_table(multiple_runs_data)
```
