```python
import json

filename = 'results/all_backbone_method_dataset_experiment_results.json'

with open(filename) as f:
    paper_results_data = json.load(f)

# print(paper_results_data)
```

```python
def generate_latex_table(json_data):
    # Datasets and model sizes based on LaTeX table
    desc_gen_method = 'gpt3'
    datasets = ['imagenet', 'cub', 'pets', 'dtd', 'food101', 'places365', 'eurosat']
    # datasets = ['imagenet', 'places365' ]
    model_sizes = ['ViT-B/32', 'ViT-B/16', 'ViT-L/14']
    methods = ['clip', 'e-clip', 'd-clip', 'waffleclip', 'defntaxs']

    # Initialize a structure to store the column values
    column_values = {dataset: {model_size: [] for model_size in model_sizes} for dataset in datasets}

    # Gather values for each column
    for method in methods:
        for dataset in datasets:
            for model_size in model_sizes:
                try:
                    if method in json_data[desc_gen_method][model_size]:
                        result = json_data[desc_gen_method][model_size][method][dataset][f"{method.capitalize()} Accuracy: "]
                    else:
                        result = "-"
                except KeyError:
                    result = "-"
                if isinstance(result, float):
                    column_values[dataset][model_size].append(result)
                else:
                    column_values[dataset][model_size].append(None)

    # Table header formatting
    print("\\begin{table}[ht]")
    print("    \\caption{Comparison of zero-shot visual classification performance across different image classification benchmarks using multiple CLIP backbones.}")
    print("    \\label{tab:clip_comparison}")
    print("    \\begin{center}")
    print("    \\resizebox{\\textwidth}{!}{")
    print("    \\begin{tabular}{lccc|ccc|ccc|ccc|ccc|ccc|ccc|ccc}")

    print("        \\hline")
    print("        \\multirow{2}{*}{\\textbf{Method}} & \\multicolumn{3}{c}{\\textbf{ImageNet}} & \\multicolumn{3}{c}{\\textbf{CUB}} & \\multicolumn{3}{c}{\\textbf{Oxford Pets}} & \\multicolumn{3}{c}{\\textbf{DTD}} & \\multicolumn{3}{c}{\\textbf{Food101}} & \\multicolumn{3}{c}{\\textbf{Places365}} & \\multicolumn{3}{c}{\\textbf{EuroSAT}} \\\\")
    print("        \\cmidrule(lr){2-4} \\cmidrule(lr){5-7} \\cmidrule(lr){8-10} \\cmidrule(lr){11-13} \\cmidrule(lr){14-16} \\cmidrule(lr){17-19} \\cmidrule(lr){20-22}")
    print("        & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 & B/32 & B/16 & L/14 \\\\")
    print("        \\hline")

    # Iterate over each method and print results in the LaTeX format
    for method in methods:
        row = f"        {method.capitalize()}"  # Start each row with the method name
        for dataset in datasets:
            for model_size in model_sizes:
                # Extract the accuracy for each combination
                try:
                    if method in json_data[desc_gen_method][model_size]:
                        result = json_data[desc_gen_method][model_size][method][dataset][f"{method.capitalize()} Accuracy: "]
                    else:
                        result = "-"
                except KeyError:
                    result = "-"

                if isinstance(result, float):
                    # Determine if the value is the max or second max in the column
                    col_values = column_values[dataset][model_size]
                    max_val = max(v for v in col_values if v is not None)
                    second_max_val = sorted([v for v in col_values if v is not None], reverse=True)[1] if len(col_values) > 1 else None

                    if result == max_val:
                        result_str = f"\\textbf{{{result:.2f}}}"
                    elif result == second_max_val:
                        result_str = f"\\underline{{{result:.2f}}}"
                    else:
                        result_str = f"{result:.2f}"
                else:
                    result_str = result
                row += f" & {result_str}"
        row += " \\\\"
        print(row.replace("Clip", "CLIP").replace("clip","CLIP").replace("Defntaxs", "DefNTaxS"))
    
    # Close the LaTeX table
    print("        \\hline")
    print("    \\end{tabular}")
    print("    }")
    print("    \\end{center}")
    print("\\end{table}")



generate_latex_table(paper_results_data)
```
